```python
# Importar librerías y módulo
import os
import importlib
import estereo  # Asegúrate de que estereo.py está en el mismo directorio que este notebook
importlib.reload(estereo)  # Por si haces cambios en estereo.py

# Verificar que el archivo estéreo existe
fichero_estereo = os.path.join("wav", "komm.wav")
assert os.path.exists(fichero_estereo), "No se encuentra el archivo 'komm.wav' en la carpeta 'wav/'"
print("Archivo estéreo encontrado:", fichero_estereo)

# Convertir estéreo a mono 
print("\n Generando versiones mono...")

estereo.estereo2mono(fichero_estereo, "komm_L.wav", canal=0)       # canal izquierdo
estereo.estereo2mono(fichero_estereo, "komm_R.wav", canal=1)       # canal derecho
estereo.estereo2mono(fichero_estereo, "komm_suma.wav", canal=2)    # semisuma
estereo.estereo2mono(fichero_estereo, "komm_resta.wav", canal=3)   # semidiferencia

print("Versiones mono generadas.")

# Reconstruir estéreo a partir de L y R
print("\n Reconstruyendo estéreo desde L y R...")
estereo.mono2estereo("komm_L.wav", "komm_R.wav", "komm_rec.wav")
print(" Estéreo reconstruido.")

# Codificar estéreo 
print("\n Codificando estéreo (semi-suma y semi-diferencia)...")
estereo.codEstereo(fichero_estereo, "komm_cod.wav")
print(" Codificación completa.")

# Decodificar estéreo 
print("\n Decodificando estéreo desde codificación...")
estereo.decEstereo("komm_cod.wav", "komm_dec.wav")
print(" Decodificación completa.")

# Escuchar el audio en el notebook 
from IPython.display import Audio

print("\n Escuchando algunas salidas:")
print("Original (komm.wav)")
display(Audio(fichero_estereo))

print("Estéreo reconstruido (komm_rec.wav)")
display(Audio("komm_rec.wav"))

print("Estéreo decodificado (komm_dec.wav)")
display(Audio("komm_dec.wav"))
```
